```python
import os
import pickle

# Load metadata file
output_dir = 'dataset/genea2023_dataset/trn/main-agent'
meta_file = os.path.join(output_dir, "consolidated_meta.pkl")

with open(meta_file, 'rb') as f:
    meta = pickle.load(f)
    
    skeleton_info = meta['skeleton_info']

    print("Skeleton info:")
    for key, value in skeleton_info.items():
        print(f"{key}: {value}")


# For each category, extract the bone_indices


categories = {'hands': 0.4, 'legs': 2.0, 'arms': 1.0, 'spine': 1.0}

for category, weight in categories.items():
    print(f"\nCategory: {category}")
    bone_indices = []
    for bone_name in skeleton_info['bone_categories'][category]:
        bone_indices.extend(skeleton_info['bone_to_indices'][bone_name])

    print(f"Bone indices for {category}: {bone_indices}")


# Based on the categories I want to construct torch tensor vector with the weights of the categories for each index

import torch
import numpy as np

# Initialize a tensor of zeros with the same length as the number of bones
num_features = skeleton_info['number_of_features']
weights_tensor = torch.ones(num_features)

# Assign weights based on the categories
for category, weight in categories.items():
    for bone_name in skeleton_info['bone_categories'][category]:
        bone_indices = skeleton_info['bone_to_indices'][bone_name]
        for index in bone_indices:
            weights_tensor[index] = weight


# print the weights tensor
print("\nWeights tensor:")
print(weights_tensor)
```
